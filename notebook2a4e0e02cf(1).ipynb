```python
from fastai.text.all import *
import pandas as pd
path = untar_data(URLs.WIKITEXT_TINY)
df = pd.read_csv(path/'train.csv', header=None)
dls = TextDataLoaders.from_df(df, path=path, text_col=0, is_lm=True, valid_pct=0.1)
learn = language_model_learner(dls,AWD_LSTM)
learn.fine_tune(1)
```

```python
learn.export('/kaggle/working/model.pkl')
```
